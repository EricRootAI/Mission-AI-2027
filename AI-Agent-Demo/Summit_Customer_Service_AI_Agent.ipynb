```python
import json

knowledge_base = {
    "company": {
        "name": "Summit Climate Systems",
        "industry": "Residential HVAC Services"
    },
    "business_hours": {
        "monday_friday": "7:00 AM - 7:00 PM",
        "saturday": "8:00 AM - 4:00 PM",
        "sunday": "Closed"
    },
    "service_area": [
        "North Ridge",
        "Lakeview",
        "Pine Valley",
        "Westbrook",
        "Summit County"
    ],
    "agent_restrictions": [
        "Do not provide final repair quotes.",
        "Do not diagnose equipment failures with certainty.",
        "Do not invent company policies.",
        "Do not claim appointment availability without access to scheduling data.",
        "Escalate potentially dangerous or urgent situations to a human representative."
    ]
}

print("Knowledge base loaded successfully.")
```

```python
def classify_request(customer_request):
    request = customer_request.lower()

    urgent_keywords = [
        "burning",
        "smoke",
        "sparks",
        "electrical smell",
        "carbon monoxide",
        "elderly",
        "infant",
        "82-year-old",
        "stopped working"
    ]

    pricing_keywords = [
        "exactly how much",
        "final quote",
        "replacement cost",
        "repair cost",
        "how much do you charge",
        "how much does it cost"
    ]

    scheduling_keywords = [
        "reschedule",
        "appointment availability",
        "available tomorrow",
        "book an appointment",
        "come to my house tomorrow",
        "tomorrow morning",
        "tomorrow afternoon"
    ]

    diagnosis_keywords = [
        "confirm that's the problem",
        "confirm that is the problem",
        "is my compressor broken",
        "diagnose"
    ]

    if any(keyword in request for keyword in urgent_keywords):
        return "ESCALATE"

    if any(keyword in request for keyword in pricing_keywords):
        return "ESCALATE"

    if any(keyword in request for keyword in diagnosis_keywords):
        return "ESCALATE"

    if any(keyword in request for keyword in scheduling_keywords):
        return "CLARIFY"

    return "ANSWER"
```

```python
test_requests = [
    "What areas do you service?",
    "My AC is running but the house isn't getting cold.",
    "Can I reschedule my appointment?",
    "I smell something burning coming from my HVAC unit.",
    "Can you tell me exactly how much replacing my compressor will cost?"
]

for request in test_requests:
    decision = classify_request(request)
    print(f"{decision}: {request}")
```

```python
def answer_from_knowledge_base(customer_request, knowledge_base):
    request = customer_request.lower()

    # Service area questions
    if (
        "area" in request
        or "service" in request
        or "come out to" in request
        or "serve" in request
    ):
        for area in knowledge_base["service_area"]:
            if area.lower() in request:
                return f"Yes. Summit Climate Systems serves {area}."

        areas = ", ".join(knowledge_base["service_area"])
        return f"Summit Climate Systems serves {areas}."

    # Business hours
    if "open" in request or "hours" in request:
        hours = knowledge_base["business_hours"]
        return (
            f"Monday-Friday: {hours['monday_friday']}; "
            f"Saturday: {hours['saturday']}; "
            f"Sunday: {hours['sunday']}."
        )

    # Air filter guidance
    if "filter" in request:
        return (
            "Most residential air filters should be inspected monthly. "
            "Replacement frequency depends on filter type, household conditions, "
            "and manufacturer recommendations."
        )

    # AC not cooling / warm air
    if (
        "not getting cold" in request
        or "isn't getting cold" in request
        or "is not getting cold" in request
        or "not cooling" in request
        or "feels warm" in request
        or "warm air" in request
        or "vents feel warm" in request
    ):
        return (
            "Please confirm the thermostat is set to COOL and below the current "
            "indoor temperature, check that the HVAC system has power, and inspect "
            "the air filter. If the issue continues, contact Summit Climate Systems."
        )

    # Thermostat problems
    if (
        "thermostat won't turn on" in request
        or "thermostat will not turn on" in request
        or "thermostat is blank" in request
        or "blank thermostat" in request
        or "thermostat screen is blank" in request
    ):
        return (
            "Check whether the HVAC system has electrical power. "
            "If the thermostat uses replaceable batteries, replace them. "
            "If the thermostat remains blank or will not turn on, "
            "contact Summit Climate Systems for service."
        )

    return "I do not have enough verified information to answer that request."
```

```python
for request in test_requests:
    decision = classify_request(request)

    if decision == "ANSWER":
        response = answer_from_knowledge_base(request, knowledge_base)

    elif decision == "CLARIFY":
        response = (
            "I need additional information before I can handle this request."
        )

    elif decision == "ESCALATE":
        response = (
            "This request should be reviewed by a human representative."
        )

    print(f"\nCustomer: {request}")
    print(f"Decision: {decision}")
    print(f"Response: {response}")
```

```python
intent_sentences = []
intent_labels = []

for intent, examples in intent_examples.items():
    for example in examples:
        intent_sentences.append(example)
        intent_labels.append(intent)

intent_embeddings = semantic_model.encode(
    intent_sentences,
    convert_to_tensor=True
)


def semantic_classify(customer_request, threshold=0.50):
    request_embedding = semantic_model.encode(
        customer_request,
        convert_to_tensor=True
    )

    similarities = util.cos_sim(
        request_embedding,
        intent_embeddings
    )[0]

    best_index = similarities.argmax().item()
    best_score = similarities[best_index].item()
    best_intent = intent_labels[best_index]

    if best_score < threshold:
        return "UNKNOWN", best_score

    return best_intent, best_score
```

```python
intent_examples = {
    "SERVICE_AREA": [
        "Do you provide service where I live?",
        "Do you come to my area?",
        "Is my town in your service area?"
    ],

    "BUSINESS_HOURS": [
        "When are you open?",
        "What are your business hours?",
        "Are you open this weekend?"
    ],

    "AC_NOT_COOLING": [
        "My air conditioner is not cooling.",
        "The air from my vents is warm.",
        "My house is not getting cold."
    ],

    "THERMOSTAT_PROBLEM": [
        "My thermostat will not turn on.",
        "The thermostat screen is blank.",
        "My thermostat has no power."
    ],

    "SCHEDULING": [
        "Can I schedule an appointment?",
        "Can someone come tomorrow?",
        "I need to reschedule my service visit."
    ],

    "PRICING": [
        "How much will this repair cost?",
        "What do you charge for repairs?",
        "Can you give me an exact price?"
    ],

    "DIAGNOSIS": [
        "Can you confirm what is broken?",
        "Tell me exactly what failed.",
        "Is my compressor broken?"
    ],

    "URGENT_SAFETY": [
        "There is smoke coming from my HVAC system.",
        "I smell something burning.",
        "There are sparks coming from my unit."
    ]
}
```

```python
def governed_semantic_decision(customer_request):
    intent, confidence = semantic_classify(customer_request)

    # Low-confidence semantic matches are not trusted.
    if intent == "UNKNOWN":
        return {
            "intent": intent,
            "confidence": confidence,
            "decision": "CLARIFY",
            "reason": "Semantic confidence below threshold."
        }

    # Safety-critical or restricted intents require human review.
    if intent in ["URGENT_SAFETY", "PRICING", "DIAGNOSIS"]:
        return {
            "intent": intent,
            "confidence": confidence,
            "decision": "ESCALATE",
            "reason": "Request falls within a restricted or safety-sensitive category."
        }

    # Scheduling cannot be completed without live scheduling access.
    if intent == "SCHEDULING":
        return {
            "intent": intent,
            "confidence": confidence,
            "decision": "CLARIFY",
            "reason": "Agent does not have access to live scheduling data."
        }

    # Verified informational intents may proceed.
    if intent in [
        "SERVICE_AREA",
        "BUSINESS_HOURS",
        "AC_NOT_COOLING",
        "THERMOSTAT_PROBLEM"
    ]:
        return {
            "intent": intent,
            "confidence": confidence,
            "decision": "ANSWER",
            "reason": "Intent can be handled using verified company information."
        }

    return {
        "intent": intent,
        "confidence": confidence,
        "decision": "CLARIFY",
        "reason": "Intent is not approved for autonomous handling."
    }
```

```python
governance_test_requests = [
    "Will your technicians travel to my neighborhood?",
    "What time do you guys shut down for the day?",
    "The system is blowing air but the house stays hot.",
    "The little screen on my thermostat is completely dead.",
    "Could I get somebody over here sometime tomorrow?",
    "What's this going to set me back financially?",
    "Can you tell me for sure that my compressor died?",
    "Something smells like it's scorching inside the unit."
]

print("SPRINT 1J - GOVERNED SEMANTIC PIPELINE\n")

for request in governance_test_requests:
    result = governed_semantic_decision(request)

    print(f"Customer: {request}")
    print(f"Intent: {result['intent']}")
    print(f"Confidence: {result['confidence']:.3f}")
    print(f"Decision: {result['decision']}")
    print(f"Reason: {result['reason']}")
    print("-" * 70)
```

```python
def governed_agent_response(customer_request, knowledge_base):
    governance = governed_semantic_decision(customer_request)

    intent = governance["intent"]
    decision = governance["decision"]

    if decision == "ESCALATE":
        if intent == "URGENT_SAFETY":
            response = (
                "This may involve a safety concern. Please stop using the HVAC "
                "system if it is safe to do so and contact Summit Climate Systems "
                "or an appropriate emergency service for assistance."
            )
        else:
            response = (
                "I can't safely provide a definitive answer to that request. "
                "A Summit Climate Systems team member should assist you."
            )

    elif decision == "CLARIFY":
        if intent == "SCHEDULING":
            response = (
                "I can help with general service information, but I don't have "
                "access to live appointment availability. A team member will need "
                "to assist with scheduling."
            )
        else:
            response = (
                "I don't have enough confidence to answer that request from "
                "verified information. Could you provide a little more detail?"
            )

    elif decision == "ANSWER":
        response = answer_from_knowledge_base(
            customer_request,
            knowledge_base
        )

    else:
        response = (
            "I don't have enough verified information to answer that request."
        )

    return {
        "intent": intent,
        "confidence": governance["confidence"],
        "decision": decision,
        "response": response
    }
```
